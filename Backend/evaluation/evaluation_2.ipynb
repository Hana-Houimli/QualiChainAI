```python
import sys
import os


sys.path.append(os.path.abspath(".."))
```

```python
from evaluate import evaluate
from test_cases import regulatory_cases ,audit_cases , capa_cases, report_cases
from graph.workflow import app
from langchain_core.messages import HumanMessage
```

```python
results_1 = []
for case in regulatory_cases:

    # 1. Exécuter le workflow
    answer = app.invoke({
        "messages": [
            HumanMessage(content=case["question"])
        ]
    })

    # 2. Évaluer avec le LLM Judge
    evaluation = evaluate(case, answer)

    # 3. Stocker le résultat
    results_1.append({
        "question": case["question"],
        "result" : answer,
        "evaluation": evaluation

    })


for result in results_1:
    print("=" * 80)
    print("QUESTION :", result["question"])
    print("ÉVALUATION :")
    print(result["evaluation"])
```
