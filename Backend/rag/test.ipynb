```python
from extraction import load_pdf
from chunking import chunk_documents
from config import *
from embedding import get_embedding_model
from retriever import Retriever
from vectordb import VectorDB
from generator import Generator
from dotenv import load_dotenv
import os
from pipeline import RAGPipeline

load_dotenv()
HF_TOKEN = os.getenv("HF_TOKEN")
```

```python
docs = load_pdf("bdp.pdf")
chunks = chunk_documents(docs,CHUNK_SIZE, CHUNK_OVERLAP)
embedding = get_embedding_model(EMBEDDING_MODEL)
```

```python
db = VectorDB(COLLECTION_NAME, embedding, PERSIST_DIRECTORY)
db.add_documents(chunks)

print("Database created successfully ! ")
```

```python
retriever = Retriever(db, TOP_K)
docs = retriever.retrieve(
    "What are GDP storage requirements?"
)

print("retrierved documents ")

for doc in docs:
    print(doc.page_content)
```

```python
print("generated response \n \n")


generator = Generator(
    model_name=LLM_MODEL,
    temperature=TEMPERATURE,
    max_new_tokens=MAX_NEW_TOKENS
)
```

```python
pipeline = RAGPipeline(
    retriever,
    generator,
)

response = pipeline.answer(
    "What are GDP storage requirements?"
)

print(response["answer"])
print(response["sources"])
```
